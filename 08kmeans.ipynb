```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

iri = load_iris()

dat = pd.DataFrame(
    iri.data,
    columns=iri.feature_names
)

print("\nDataset:")
print(dat.head())

print("\nDataset Shape:")
print(dat.shape)

print("\nDataset Information:")
dat.info()

print("\nStatistical Summary:")
print(dat.describe())

print("\nMissing Values:")
print(dat.isnull().sum().sum())

print("\nDuplicate Values:")
print(dat.duplicated().sum())

x = dat.copy()

sca = StandardScaler()

xsca = sca.fit_transform(x)

iner = []

for k in range(1, 11):

    mod = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    mod.fit(xsca)

    iner.append(mod.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    iner,
    marker='o',
    color='blue'
)

plt.xlabel('number of clusters')
plt.ylabel('inertia')
plt.title('elbow method')
plt.grid(True)
plt.show()

mod = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

ypred = mod.fit_predict(xsca)

dat['cluster'] = ypred

sil = silhouette_score(xsca, ypred)

print("\n========== MODEL EVALUATION ==========")

print("Number of Clusters:", 3)
print("Inertia:", mod.inertia_)
print("Silhouette Score:", sil)

print("\nCluster Distribution:")
print(dat['cluster'].value_counts())

print("\nCluster Centers:")
print(sca.inverse_transform(mod.cluster_centers_))

print("\nClustered Dataset:")
print(dat.head(10))

pca = PCA(n_components=2)

xpc = pca.fit_transform(xsca)

vis = pd.DataFrame({
    'pc1': xpc[:, 0],
    'pc2': xpc[:, 1],
    'cluster': ypred
})

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=vis,
    x='pc1',
    y='pc2',
    hue='cluster',
    palette='viridis',
    s=70
)

plt.title('k-means clustering')
plt.xlabel('principal component 1')
plt.ylabel('principal component 2')
plt.legend(title='cluster')
plt.show()

plt.figure(figsize=(8, 5))

sns.countplot(
    data=dat,
    x='cluster',
    hue='cluster',
    palette='viridis',
    legend=False
)

plt.title('cluster distribution')
plt.xlabel('cluster')
plt.ylabel('count')
plt.show()

print("\n========== CLUSTER ANALYSIS ==========")

print(
    dat.groupby('cluster').mean(numeric_only=True)
)

print("\n========== PROGRAM COMPLETED ==========")
```
