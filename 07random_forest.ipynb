```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import precision_score, recall_score, f1_score

can = load_breast_cancer()

dat = pd.DataFrame(
    can.data,
    columns=can.feature_names
)

dat['target'] = can.target

print("\nDataset:")
print(dat.head())

print("\nDataset Shape:")
print(dat.shape)

print("\nDataset Information:")
dat.info()

print("\nStatistical Summary:")
print(dat.describe())

print("\nMissing Values:")
print(dat.isnull().sum().sum())

print("\nDuplicate Values:")
print(dat.duplicated().sum())

x = dat.drop('target', axis=1)
y = dat['target']

xtr, xte, ytr, yte = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

mod = RandomForestClassifier(
    n_estimators=100,
    criterion='gini',
    max_depth=None,
    random_state=42
)

mod.fit(xtr, ytr)

ypred = mod.predict(xte)

prob = mod.predict_proba(xte)

acc = accuracy_score(yte, ypred)
pre = precision_score(yte, ypred)
rec = recall_score(yte, ypred)
f1 = f1_score(yte, ypred)

cm = confusion_matrix(yte, ypred)

print("\n========== MODEL EVALUATION ==========")

print("Accuracy:", acc)
print("Precision:", pre)
print("Recall:", rec)
print("F1 Score:", f1)

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(
    yte,
    ypred,
    target_names=can.target_names
))

res = pd.DataFrame({
    'actual': yte.values,
    'predicted': ypred
})

print("\nActual vs Predicted:")
print(res.head(10))

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=can.target_names,
    yticklabels=can.target_names
)

plt.xlabel('predicted')
plt.ylabel('actual')
plt.title('random forest confusion matrix')
plt.show()

plt.figure(figsize=(8, 5))

sns.countplot(
    x=y,
    hue=y,
    palette='viridis',
    legend=False
)

plt.xticks(
    [0, 1],
    can.target_names
)

plt.xlabel('tumor type')
plt.ylabel('count')
plt.title('class distribution')
plt.show()

imp = pd.DataFrame({
    'feature': x.columns,
    'importance': mod.feature_importances_
})

imp = imp.sort_values(
    by='importance',
    ascending=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=imp.head(10),
    x='importance',
    y='feature',
    hue='feature',
    palette='viridis',
    legend=False
)

plt.title('top 10 feature importance')
plt.xlabel('importance')
plt.ylabel('features')
plt.tight_layout()
plt.show()

new = pd.DataFrame(
    [can.data[0]],
    columns=can.feature_names
)

pre = mod.predict(new)

print("\n========== NEW PREDICTION ==========")

print("Predicted Class:", can.target_names[pre[0]])

print("\nPrediction Probability:")
print(mod.predict_proba(new))

print("\n========== PROGRAM COMPLETED ==========")
```
